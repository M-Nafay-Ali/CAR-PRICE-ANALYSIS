This is to import our Car dataset and show how much missing values are present or is all the numerical values are in integat or float data type not in strings
In short to grasp the insight of raw data

```python
import pandas as pd
Path='/content/All Brand Cars price in Pakistan.csv'
df=pd.read_csv(Path)
print(df)
print("-----------")
print(df.head())
print("-------------")
print(df.info())
print("________________________")

# Removing commas and converting to float
```

This step is to clean data
•Handling missing values
•converting string datatype into float data type
Then printing them to check if it is processed without any problem or error

```python
df['Min Price'] = df['Min Price'].astype(str).str.replace(',', '').astype(float)
df['Max Price'] = df['Max Price'].astype(str).str.replace(',', '').astype(float)

print("Missing values in each column:")
print(df.isnull().sum())


print(df.dtypes)
```

In this we are going to
•Create new column in our data set called "Average" where as it's name suggest it will be showing average price of cars
And make our data sort in ascending order then create a graph to visualize our data
Note: the prices are in exponent values as 10³ etc

```python
df['Avg Price'] = (df['Min Price'] + df['Max Price']) / 2

df.groupby('Make')['Avg Price'].mean().sort_values(ascending=False).head(10)

df.groupby('Year')['Avg Price'].mean().plot(kind='line')

import plotly.express as px


fig = px.scatter(df, x="Year", y="Avg Price", color="Make", hover_data=['Model'])
fig.show()


print(df[['Min Price', 'Max Price', 'Year']].describe())
```
